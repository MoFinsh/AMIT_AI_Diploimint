```python
import numpy as np
# NumPy: library for handling numbers, arrays, and mathematical operations
import matplotlib.pyplot as plt
from matplotlib.pyplot import axes
from pandas.core.common import random_state
# Matplotlib: library for creating visualizations and plots
from sklearn.datasets import make_moons
# make_moons: function to create a synthetic dataset shaped like two "moons"
# (commonly used to test clustering algorithms)
from sklearn.cluster import KMeans, DBSCAN
# KMeans: clustering algorithm that divides data into k groups
# DBSCAN: clustering algorithm based on density (can find noise/outliers too)
```

```python
X, y = make_moons(n_samples=400, noise=0.1, random_state=42)
```

```python
# Apply KMeans
kmeans= KMeans(n_clusters=3, random_state=42)
```

```python
y_kmeans = kmeans.fit_predict(X)
```

```python
# Apply DBSCAN
dbscan = DBSCAN(eps=0.2, min_samples=5)
```

```python
y_dbscan = dbscan.fit_predict(X)
```

```python
# -------------------------------------------------
# 4) Plot results for comparison
# -------------------------------------------------

# Create a figure with 1 row and 2 columns of subplots
# - fig → the overall figure
# - axes → an array holding each subplot (axes[0] and axes[1])
# - figsize=(12, 5) → width=12 inches, height=5 inches
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# -------------------------
# Left Plot → K-Means result
# -------------------------
# scatter() → plot the points
# X[:, 0] = x-coordinates, X[:, 1] = y-coordinates
# c=y_kmeans → color each point by its assigned cluster label
# cmap="viridis" → use the 'viridis' color map for coloring
axes[0].scatter(X[:, 0], X[:, 1], c=y_kmeans, cmap="viridis")
axes[0].set_title("K-Means Clustering")  # Title for the subplot

# -------------------------
# Right Plot → DBSCAN result
# -------------------------
# Same scatter plot, but colored by DBSCAN cluster labels
# DBSCAN may also assign -1 (colored differently) for noise/outliers
axes[1].scatter(X[:, 0], X[:, 1], c=y_dbscan, cmap="viridis")
axes[1].set_title("DBSCAN Clustering")  # Title for the subplot

# -------------------------
# Display both plots side by side
# -------------------------
plt.show()
```
