```python
class FLiers :
    def __init__(self,wing_size, name = ""):
        self.name = name
        self.wing_size = wing_size

    def fly (self):
        return f"the {self.name} can fly and wing size is {self.wing_size}"

class swimmer :
    def __init__(self, fin_size, name = ''):
        self.name = name
        self.fin_size = fin_size

    def swim(self):
        return f"the {self.name} can swim and fin size is {self.fin_size}"
    
class both(FLiers,swimmer) :
    def __init__(self, name, wing_size, fin_size):  
        FLiers.__init__(self,wing_size)
        swimmer.__init__(self, fin_size)
        self.name = name



    def get_info(self):
        print(f"the {self.name} can fly the wing size is : {self.wing_size} and swimming the fin size is {self.fin_size}")
    


duck = both("Duckk",20,10)
print(duck.fly())
print(duck.swim())
duck.get_info()
```

```python
class Child:
    def __init__(self, age, school):
        self.age = age
        self.school = school

    def info(self):
        print("Iam a child")
    
class Student(Child):
    def __init__(self, age, school, grade):
        super().__init__(age, school)
        self.grade = grade
        
    def info(self):
        print("Iam a student")

student_1 = Student(20, "AMIT", 95)
student_1.info()
Child.info(student_1)
```

```python
class Animal:
    def __init__(self,name):
        self.name = name
    
    def info(self):
        return f"{self.name} is an animal"

class Mammal(Animal):
    def __init__(self, name):
        super().__init__(name)
    
    def Eat(self):
        print(f"This {self.name} eats")

    def info(self):
        info2 = super().info()
        print( f"{info2}  y name is mammal ")

class Dog(Mammal):
    def __init__(self, name):
        super().__init__(name)

    def walk(self):
        print("Iam walk")

    def info(self):
        info3 = super().info()
        print(f"{info3} my name is dog ")

dog_1 = Dog("Ricko")
dog_1.Eat()
dog_1.info()
```

```python
class Mobile:
    def __init__(self, brand):
        self.brand = brand

    def Power_on(self):
        print("mobile is running ")
    
    def Info(self):
        return self.brand

class Phone(Mobile):
    def __init__(self, brand, number):
        Mobile.__init__(self, brand)
        self.number = number       

    def mack_a_call(self):
        print("Calling")

class Camera(Mobile):
    def __init__(self, brand, pixel):
        self.pixel = pixel
        Mobile.__init__(self, brand)

    def Tack_a_photo(self):
        print("photo is tacken")

class Smart_phone(Phone,Camera):
    def __init__(self, brand, number, pixel):
        Camera.__init__(self, brand, pixel)
        Phone.__init__(self, brand, number)
    def Info(self):
        string = super().Info()
        print(f"{string} and the mobile number is {self.number} and the camera is {self.pixel} mega pixel")

Phone = Smart_phone("iphone","01271293727",8000)
Phone.Tack_a_photo()
Phone.mack_a_call()
Phone.Info()


```
