```python
class Students:
    students_num = 0
    
    def __init__(self,name,age,*courses):
        self.__name = name
        self.__age = age
        self.__course = courses
        Students.students_num +=1

    def set_name(self, new_name):       #To set new value to my object
        self.__name = new_name

    def get_name(self):                #to get my value to see it 
        return self.__name

    def Info(self):
        return f"hello {self.__name} your age is : {self.__age} your courses is : {self.__course}\n"

print(Students.students_num)
student_1 = Students("mohamed",20,"CSS","Java script")

student_1.set_name("ali")                   #the only way to change data in cabsulition (privsy)

student_2 = Students("Ahmed",25,"HTML","Java ")
print(Students.students_num)

print(f"{student_1.Info()}{student_2.Info()}")
```

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Person:
    def __init__(self,name,age,number):
        self.name = name
        age = int(age)
        self.age = age
        # number = int(number)
        self.number = number
    
    def display(self):
        return f"hello {self.name}, your age is {self.age}, your number is {self.number} "
    
    @classmethod
    def InitBirthYear(cls,name,birthYear,number = "None",extra = "none"):
        birthYear = int(birthYear)
        return cls(name ,date.today().year - birthYear , number, extra)
        
class Square(Shape):
    def __init__(self,side):
        self.side = side

    def area(self):
        return self.side*self.side
    
    def perimeter(self):
        return self.side * 4
    
class Rectingle(Shape):
    def __init__(self,hight,width):
        self.hight = hight
        self.width = width
    
    def area(self):
        return self.hight * self.width
    
    def perimeter(self):
        return self.hight + self.width * 2
    
square = Square(10)
print(f"square area is {square.area()} and perimeter is {square.perimeter()}")
rectingle = Rectingle(5,3)
print(f"rectingle area is {rectingle.area()} and perimeter is {rectingle.perimeter()}")
```
