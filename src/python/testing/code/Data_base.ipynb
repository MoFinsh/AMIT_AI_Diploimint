```python
import pyodbc


conn_str = (
    'DRIVER={ODBC Driver 17 for SQL Server};'
    'SERVER=localhost\\SQLEXPRESS;'
    'DATABASE=Company;'
    'Trusted_Connection=yes;'
)
def get_db_connection():
    connection = pyodbc.connect(conn_str)
    return connection
```

```python
def update_customer(customer_id):
    try:
        conn = get_db_connection()
        cursor = conn.cursor()

        sql_select ="""SELECT Customer_ID, Customer_Name, Age, Gender, Region
                    FROM dbo.Customers
                    WHERE Customer_ID= ?"""
        cursor.execute(sql_select,customer_id)
        row =  cursor.fetchone()
        if row is None:
            return False, "Customer not found"
        print("Current data")
        print(row)

        new_name = input("Enter new name: ")
        new_age = int(input("Enter new age: "))
        new_gender = input("Enter new gender: ")
        new_region = input("Enter new region: ")
        sql_update = """ UPDATE dbo.Customers
                  SET Customer_Name=?, Age=?, Gender=?, Region=?
                  WHERE Customer_ID= ? """

        cursor.execute(sql_update, (new_name, new_age, new_gender, new_region, customer_id))
        cursor.execute(sql_select,customer_id)
        row =  cursor.fetchone()
        print(row)
        conn.commit()
        return True, f"Customer updated successfully"

    except Exception as e:
        return False, str(e)
    finally:
        conn.close()

```

```python
update_customer(2)
```

```python
def delete_customer(customer_id):
    try:
        conn = get_db_connection()
        cursor = conn.cursor()

        sql_delete = """DELETE FROM dbo.Customers WHERE Customer_ID= ?"""
        cursor.execute(sql_delete,customer_id)
        cursor.commit()
        return True, f"Customer deleted successfully"
    except Exception as e:
        return False, str(e)
```

```python
delete_customer(30)
```

# add many in mongo db

```python
import pymongo
from pymongo import MongoClient
from bson.objectid import ObjectId

client = MongoClient("mongodb://localhost:27017/")
db = client["CustomersDB"]
customers_col = db["Customers"]
```

```python
def add_many_customer(customer):
    try:
        result = customers_col.insert_many(customer)
        return True, str(result.inserted_ids)
    except Exception as e:
        return False, str(e)
```

# delete one in mongo db

```python
def delete_customer(custmer_id):
    try:
        result = customers_col.delete_one({"_id": ObjectId(custmer_id)})
        return True, f"Customer deleted successfully"
    except Exception as e:
        return False, str(e)
```

# delete many in mongo db

```python
def delete_many_customer(customer_name):
    try:
        result = customers_col.delete_many({"name": customer_name})
        return True, f"Customer deleted successfully"
    except Exception as e:
        return False, str(e)
```
