```python
task2 = "###!!@mocleW EPGTQ!!!6789"
new_task2 = task2[6:-7:1]
print(f"first step is :{new_task2}")    #first step
print(f"second step is :{new_task2.replace("mocleW","Welcome")}")     #second step

# رغم انك مش طالب تعديل ف الكلمه الثانية بس جبتلك الطلب الي انت عايزو 😉😂

print(f"second step is :{new_task2.replace("EPGTQ","PGTQ")}")         #final step
```

```python
task3 = "&&&**$gnirtS PLIO!!@1234"
new_task3 = task3[6:-7:1]
print(f"first step is :{new_task3}")   #first step

first_word_str_to_list = list(new_task3[0:6:1])    #editing the first word 
first_word_str_to_list.reverse()
first_word_list_to_str = '' .join(first_word_str_to_list)

second_word_str_to_list = list(task3[12:17:1])   #editing the second word
second_word_str_to_list[3] = "E"
second_word_str_to_list[4] = "U"
second_word_list_to_str = '' .join(second_word_str_to_list)

final_word = first_word_list_to_str+second_word_list_to_str #final word

print(f"first word after editing : {first_word_list_to_str}")
print(f"second word after editing : {second_word_list_to_str}")
print(f"final word is : {final_word}")
```

```python
task4 = "##$$$@!yalpstcejorp EPUVT****9887"
new_task4 = task4[7:-8:1]
print(f"furst step is :{new_task4}")

first_word_str_to_list = list(new_task4[0:12:1])    #editing the first word 
first_word_str_to_list.reverse()
first_word_list_to_str = '' .join(first_word_str_to_list)

second_word_str_to_list = list(new_task4[12:18:1])   #editing the second word
second_word_str_to_list[1] = "A"
second_word_str_to_list[3] = "O"
second_word_list_to_str = '' .join(second_word_str_to_list)

final_word = first_word_list_to_str+second_word_list_to_str #final word

print(f"first word after editing : {first_word_list_to_str}")
print(f"second word after editing : {second_word_list_to_str}")
print(f"final word is : {final_word}")

#خدت بالي يهندسه ان التاسك الخير الطلب النهائي غلط فوق الحل الصح والي جاي الطلب الغلط الي كنت عايزو احنا مش قليلين بردو 😉😂
second_word_str_to_list[3] = "T"
second_word_str_to_list[4] = "O"
second_word_str_to_list[5] = "V"
wrong_answer = ''.join(second_word_str_to_list)
print(first_word_list_to_str+wrong_answer)
```
