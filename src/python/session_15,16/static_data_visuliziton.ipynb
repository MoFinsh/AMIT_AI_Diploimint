```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
days = ["Sat", "Sun", "Mon", "Tue", "Wed", "Thu", "Fri"]
cairo_temps = [30, 32, 31, 29, 28, 27, 26]
alex_temps = [25, 26, 27, 28, 28, 29, 30]
sharm_temps = [33, 33, 34, 30, 29, 28, 27]
```

```python
#Create figure (canvas)
plt.figure(figsize=(8,5))
```

```python
# Draw plot lines
sns.set_style("whitegrid")
plt.plot(days, cairo_temps, label="Cairo", color="orange", linewidth=2)
plt.plot(days, alex_temps, label = "Alex", color = "blue", linestyle = "--", linewidth = 2)
plt.plot(days, sharm_temps, label = "Sharm", color = "red", linewidth = 5)

plt.title("Weekly Temerature Comparison between Cities")
plt.xlabel("Days of the Week")
plt.ylabel("Temerature")
plt.legend()
plt.grid(True)
```

```python
# Bar chart
# Define category names
products = ["TV", "Mobile", "AC", "Fan"]
# Define values for each category
quantities = [5, 7, 3, 4]
plt.figure(figsize=(6,4))
sns.set_style("white")
plt.bar(products, quantities, color=["skyblue", "salmon", "lightgreen", "blue"])

plt.title("Products Quntities")
plt.xlabel("Products")
plt.ylabel("Quantities")
```

```python
#scattra graph
np.random.seed(0)
x_scatter = np.random.rand(50)
y_scatter = np.random.rand(50)
```

```python
plt.figure(figsize=(6,4))
plt.scatter(x_scatter, y_scatter, color="red", alpha=0.7)
plt.title("Scatter Plot Example")
plt.xlabel("Random X")
plt.ylabel("Random Y")
```

```python
# Histogram
data = np.random.randn(1000)
plt.figure(figsize=(6,4))
plt.hist(data, bins = 30, color="steelblue", edgecolor="black")
plt.title("Histogram Example")
```

```python
quantities = [30, 20, 25, 25]
fruits = ["Apples", "bananas", "cherries", "Dates"]
plt.figure(figsize=(6,6))
plt.pie(quantities, labels=fruits, autopct = "%1.1f%%", colors = sns.color_palette("pastel"))
plt.title("Pie Plot Example")
```

```python
# Load tips dataset
tips = sns.load_dataset("tips")
tips
```

```python
plt.figure(figsize=(6,4))
sns.countplot(
    x = "day",
    data = tips,
    hue = "smoker",
    palette="viridis",
    legend=False
)
```

```python
# Box blot
plt.figure(figsize=(6,4))
sns.boxplot(x= "day", y="total_bill", data=tips, palette="viridis")
```

```python
# Heat map
numeric_curr = tips.corr(numeric_only=True)
plt.figure(figsize=(8,5))
sns.heatmap(numeric_curr, annot=True, cmap="YlGnBu")
plt.show()
```
