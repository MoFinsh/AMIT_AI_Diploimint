# functions practices

1) even & odd counter

```python
def even_odd_counter(my_list):
    
    even_list = []
    odd_list = []
    even_count = 0
    odd_count = 0

    for y in my_list:
        if y %2 == 0:
            even_count+=1
            even_list.append(y)
        else:
            odd_count+=1
            odd_list.append(y)

    print(f"even list is : {even_list}")
    print(f"odd list is :{odd_list}")
    return f"even numbers count is : {even_count} , odd numbers count is : {odd_count}"

# call the function
my_list = [1,2,3,4,5,6,7,8,9]
even_odd_counter(my_list)
```

2) Lucky Number Checker

```python
def lucky_number(list):


    print(50*"*","\nthe lucky number is ends with a 7 \n and divisible by 7\n",50*"*")

    lucky_numbers =[]
    for i in list:
        if i%7 ==0 :
            m = str(i)
            if m.endswith("7"):
                lucky_numbers.append(i)

    return f"lucky numbers is : {lucky_numbers}"

# call function
my_list = [27, 7, 77,777,50]
lucky_number(my_list)
```

3) word length sorter

```python
def word_length(num):
    num.sort(key=len)
    for i in num:
        if len(i)<5:
            print(f"'{i}' ((is short word)).".center(50,'*'))
        else:
            print(f"'{i}' ((is long word)).".center(50,'*'))
    return num

# call function
num = ["hello", "python", "yes", "hi", "learning"]
word_length(num)
```

 4) Find Largest Number (Using Loop + If)

```python
def find_max(nums):
    max_num = nums[0]
    for x in nums:
        if x>max_num:
            max_num =x
            
    return f" max number is : ( {max_num} ) "

# call function
my_list = [27, 7, 77,777,50]
find_max(my_list)
```

5) Number Range Classifier (Harder Logic)


```python
def classify_number(n):
    list_1 = list(range(1,n+1))
    for i in list_1:

        if i%3 ==0 and i%5 ==0:
         print(f"'( {i} )' (( FizzBuzz ))")
        elif i %3 ==0 :
            print(f"'( {i} )' (( Fizz ))")
        elif i%5 ==0 :
            print(f"'( {i} )' (( Buzz ))")
        else:
            print(f"'( {i} )'")

# call the function
classify_number(15)
```
